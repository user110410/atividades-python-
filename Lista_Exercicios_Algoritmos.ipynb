```python
#questao_2
a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))

if a > b:
    print("Maior:", a)
elif b > a:
    print("Maior:", b)
else:
    print("Os números são iguais.")
```

```python
#questao_4
numero = int(input("Digite um número inteiro: "))
if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

```python
#questao_5
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2
print(f"Média calculada: {media:.2f}")

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")
```

```python
#questao_8
N = int(input("Digite N: "))
soma = 0
for i in range(1, N + 1):
    soma += i

print("Soma:", soma)
```

```python
#questao_9
qtd_pares = 0
for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    if numero % 2 == 0:
        qtd_pares += 1

print(f"Foram digitados {qtd_pares} números pares.")
```

```python
#questao_10
soma_positivos = 0
for i in range(8):
    numero = int(input(f"Digite o {i+1}º número: "))
    if numero > 0:
        soma_positivos += numero

print("Soma dos positivos:", soma_positivos)
```

```python
#questao_11
primeiro = int(input("Digite o 1º número: "))
maior = primeiro

for i in range(2, 7):
    numero = int(input(f"Digite o {i}º número: "))
    if numero > maior:
        maior = numero

print("Maior valor:", maior)
```

```python
#questao_12
numero = int(input("Digite um número para ver a tabuada: "))
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
#questao_13
SENHA_C = "1234"
senha = input("Digite a senha: ")

while senha != SENHA_C:
    senha = input("Senha incorreta. Tente novamente: ")

print("Acesso permitido!")
```

```python
#questao_14
NUMERO_S = 7
tentativas = 0

chute = int(input("Tente descobrir o número secreto: "))
tentativas += 1

while chute != NUMERO_S:
    if chute < NUMERO_S:
        print("Tente um número maior")
    else:
        print("Tente um número menor")

    chute = int(input("Digite outro número: "))
    tentativas += 1

print("Parabéns! Você acertou.")
print(f"Tentativas necessárias: {tentativas}")
```

```python
#questao_15
10
q_total = 0
q_positivos = 0
q_negativos = 0
soma = 0

while True:
    numero = int(input("Digite um número (ou 0 para encerrar): "))
    if numero == 0:
        break

    q_total += 1
    soma += numero

    if numero > 0:
        q_positivos += 1
    elif numero < 0:
        q_negativos += 1

print(f"Quantidade: {q_total}")
print(f"Positivos: {q_positivos}")
print(f"Negativos: {q_negativos}")
print(f"Soma: {soma}")
```
